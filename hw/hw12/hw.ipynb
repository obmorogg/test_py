# Домашнее задание. Вычисления и массивы с NumPy

## Общие правила

Во всех заданиях используйте **NumPy и только возможности NumPy, разобранные на занятии**.

Не используйте для вычислений:

```python
for
while
list comprehension
sum()
min()
max()
```

из стандартного Python, если это можно выполнить средствами NumPy.

Не записывайте вручную промежуточные и итоговые массивы, которые должны быть получены вычислением.

Например, если необходимо получить:

```text
50, 100, 150, 200, 250, 300
```

нельзя писать:

```python
np.array([50, 100, 150, 200, 250, 300])
```

Такой массив нужно сформировать средствами NumPy.

Исходные данные, которые уже даны в условии задачи, можно передавать в `np.array()` напрямую.

# Задание 1. Данные датчиков

Даны значения:

```python
[120, 180, 240, 300, 360]
```

Создайте из них NumPy-массив с типом:

```python
np.int16
```

Выведите:

* массив;
* количество измерений;
* форму;
* количество элементов;
* тип данных;
* размер одного элемента в байтах;
* общий объём массива в байтах.

### Пример результата

```text
[120 180 240 300 360]
Измерений: 1
Форма: (5,)
Элементов: 5
Тип: int16
Размер элемента: 2
Размер массива: 10
```

---

````python

import numpy as np

a = np.array([120, 180, 240, 300, 360], dtype=np.int16)
print(a)
print(f"Измерений: {a.ndim}")
print(f"Форма: {a.shape}")
print(f"Элементов: {a.size}")
print(f"Тип: {a.dtype}")
print(f"Размер элемента: {a.itemsize}")
print(f"Размер массива: {a.nbytes}")
````
